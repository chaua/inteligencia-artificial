
# Estratégias para Resolução de Problemas Através de Busca

Para resolver um problema utilizando técnicas de busca, é fundamental primeiro definir e modelar o problema adequadamente. Isso envolve os seguintes componentes:

*   **Definição do Problema:** A formulação clara do problema que se deseja resolver.

*   **Estado Inicial:** O ponto de partida no espaço de busca. Representa a situação atual do problema.

*   **Estado Final (Objetivo):** O estado ou conjunto de estados que representam uma solução para o problema. O objetivo da busca é alcançar um desses estados.

*   **Modelagem da Transição (Operadores/Ações):** As regras ou ações que permitem ir de um estado para outro. Cada ação tem pré-condições (para ser aplicável) e pós-condições (o resultado de sua aplicação).

*   **Custo do Caminho:** Uma função que atribui um valor numérico a um caminho (sequência de ações) desde o estado inicial até um determinado estado. O objetivo pode ser encontrar o caminho de menor custo.

*   **Ações:** As operações que podem ser realizadas para mudar o estado atual do problema. Cada ação leva a um novo estado.

## Classe  `Problema`

A classe abstrata `Problema` define a estrutura fundamental para qualquer problema de busca. Ela atua como um contrato, garantindo que todo problema de busca implementado possua os elementos essenciais para que os algoritmos de busca possam operá-lo de forma consistente e reutilizável.

### Conteúdo da Classe `Problema`:

*   **`estado_inicial`**: Propriedade que retorna o estado inicial do problema.

*   **`eh_objetivo(estado)`**: Método abstrato que verifica se um dado estado é um estado objetivo.

*   **`obter_sucessores(estado)`**: Método abstrato que, dado um estado, retorna uma lista de tuplas contendo a ação que levou ao novo estado, o novo estado e o custo da ação.

*   **`calcular_custo_caminho(custo_acumulado, custo_acao)`**: Método abstrato para calcular o custo acumulado de um caminho, somando o custo anterior com o custo da ação atual.

````python
from abc import ABC, abstractmethod
from typing import TypeVar, Generic, List, Tuple, Set, Any, Optional
from enum import Enum
from dataclasses import dataclass
````

````python
from abc import ABC, abstractmethod
from typing import TypeVar, Generic, List, Tuple, Set, Any, Optional
from enum import Enum
from dataclasses import dataclass

# Define tipos genéricos para o estado e a ação do problema
Estado = TypeVar('Estado')
Acao = TypeVar('Acao')


class Problema(Generic[Estado, Acao], ABC):
    """
    Classe abstrata base para a definição de um problema de busca.
    Define a interface que qualquer problema de busca concreto deve implementar, encapsulando os
    componentes teóricos necessários para sua formulação.
    """

    def __init__(self, estado_inicial: Estado):
        self._estado_inicial = estado_inicial


    # Correspondente ao 'Estado Inicial' dos conceitos teóricos.
    @property
    def estado_inicial(self) -> Estado:
        return self._estado_inicial


    # Correspondente a 'Estado Final (Objetivo)'
    @abstractmethod
    def eh_objetivo(self, estado: Estado) -> bool:
        ...


    # Correspondente a 'Modelagem da Transição (Operadores/Ações)'
    @abstractmethod
    def obter_sucessores(self, estado: Estado) -> List[Tuple[Acao, Estado, float]]:
        ...


    # Correspondente ao 'Custo do Caminho'
    @abstractmethod
    def calcular_custo_caminho(self, custo_acumulado_anterior: float, custo_da_acao_atual: float) -> float:
        ...
````

# Busca sem Informação

Busca sem Informação, ou busca não informada, é uma categoria de algoritmos de busca que não utiliza nenhuma informação adicional sobre a distância de um estado até o objetivo. Eles exploram o espaço de busca de forma sistemática, sem conhecimento heurístico para guiar a direção, baseando-se apenas na estrutura do problema.

## Tipos de Busca sem Informação

A classe `Busca` que implementamos é genérica e pode executar diferentes estratégias de busca não informada, dependendo de como a `Fronteira` é gerenciada. Os principais tipos são:

1.  **Busca em Largura (Breadth-First Search - BFS)**:
    *   **Funcionamento**: Explora o espaço de busca nível por nível, expandindo todos os nós em uma profundidade antes de passar para a próxima. Garante encontrar o caminho mais curto em termos de número de ações.
    *   **Fronteira**: Implementada como uma **fila (FIFO - First-In, First-Out)**. O nó mais antigo (adicionado primeiro) é o próximo a ser expandido.

2.  **Busca em Profundidade (Depth-First Search - DFS)**:
    *   **Funcionamento**: Explora o espaço de busca o mais profundamente possível ao longo de cada ramo antes de retroceder. Pode encontrar soluções rapidamente se o caminho for profundo, mas não garante a otimalidade (caminho mais curto).
    *   **Fronteira**: Implementada como uma **pilha (LIFO - Last-In, First-Out)**. O nó mais recente (adicionado por último) é o próximo a ser expandido.

3.  **Busca de Custo Uniforme (Uniform Cost Search - UCS)**:
    *   **Funcionamento**: Similar à Busca em Largura, mas expande o nó com o menor custo de caminho acumulado do início. Garante encontrar o caminho de menor custo total, mesmo que as ações tenham custos variados.
    *   **Fronteira**: Implementada como uma **fila de prioridade**, onde a prioridade é o custo acumulado do caminho. O nó com o menor custo total é o próximo a ser expandido.

4.  **Busca em Profundidade Limitada (Depth-Limited Search - DLS)**:
    *   **Funcionamento**: Uma variação da Busca em Profundidade que impede a busca de ir além de uma profundidade pré-definida `L`. Se uma solução não for encontrada dentro do limite `L`, a busca falha. É útil quando se sabe que a solução está dentro de uma certa profundidade.
    *   **Fronteira**: Também usa uma **pilha (LIFO)**, mas com uma verificação adicional de profundidade.

5.  **Busca em Profundidade Iterativa (Iterative Deepening Depth-First Search - IDDFS)**:
    *   **Funcionamento**: Combina as vantagens da BFS e DFS. Realiza sucessivas buscas em profundidade limitada, com limites de profundidade crescentes (0, 1, 2, ...). É completa e ótima para custos uniformes (como BFS), e usa menos memória que BFS (como DFS).
    *   **Fronteira**: Em cada iteração, usa uma **pilha (LIFO)** para a busca em profundidade limitada.

````python
class TipoBusca(Enum):
    BuscaEmLargura = "Busca em Largura"
    BuscaEmProfundidade = "Busca em Profundidade"
    BuscaDeCustoUniforme = "Busca de Custo Uniforme"
````

````python
@dataclass
class Nodo:
    estado: Estado # O estado do problema representado por este nó
    pai: Optional['Nodo'] = None # Referência ao nó pai na árvore de busca (para reconstrução do caminho)
    acao: Optional[Acao] = None # Ação que levou do nó pai a este nó
    custo_caminho: float = 0.0 # Custo acumulado do caminho desde o nó inicial até este nó
    profundidade: int = 0 # Profundidade deste nó na árvore de busca (número de ações do início)

    def __lt__(self, other):
        """
        Define a comparação 'menor que' para uso em filas de prioridade (heap).
        Compara com base no custo do caminho.
        """
        return self.custo_caminho < other.custo_caminho
````

````python
class Fronteira:
    """
    Classe para gerenciar a fronteira da busca, encapsulando a lógica de adição e remoção de nós para
    diferentes tipos de busca.
    """

    def __init__(self, tipo_busca: TipoBusca):
        self.tipo_busca = tipo_busca
        self._container = []


    def adicionar(self, nodo: Nodo):
        """
        Adiciona um nó à fronteira com base no tipo de busca.
        """
        self._container.append(nodo)


    def remover(self) -> Nodo:
        """
        Remove e retorna o nó mais prioritário da fronteira com base no tipo de busca.
        """
        if self.esta_vazia():
            raise IndexError("A fronteira está vazia.")

        # Usa um match statement para escolher a estratégia de remoção com base no tipo de busca
        match self.tipo_busca:

            # Remove o primeiro elemento (FIFO) - maior prioridade é o mais antigo
            case TipoBusca.BuscaEmLargura:
                return self._container.pop(0)

            # Remove o último elemento (LIFO) - maior prioridade é o mais recente
            case TipoBusca.BuscaEmProfundidade:
                return self._container.pop()

            # Para Busca de custo uniforme, a maior prioridade é o nó com o MENOR custo de caminho.
            # Encontra o índice do nó com o menor custo de caminho e o remove.
            case TipoBusca.BuscaDeCustoUniforme:
                return self._container.pop(min(enumerate(self._container), key=lambda item: item[1].custo_caminho)[0])

            case _:
                raise ValueError(f"Tipo de busca não reconhecido: {self.tipo_busca}")


    def esta_vazia(self) -> bool:
        return not self._container
````

## Algoritmo Geral de Busca em Árvore

Este pseudocódigo é uma representação de um algoritmo de busca geral baseado em árvore:

```
função BUSCA-EM-ARVORE(problema, fronteira):
    fronteira ← inserir(criar_nó(estado_inicial(problema)), fronteira)

    enquanto fronteira ≠ ∅ faça

1.   Item da lista
2.   Item da lista


        nó ← remover_da_fronteira(fronteira)

        se teste_objetivo(estado(nó)) então
            retornar solução(nó)

        filhos ← expandir(nó, problema)
        fronteira ← inserir_todos(filhos, fronteira)

    retornar falha
```
```
função EXPANDIR(nó, problema):
    sucessores ← ∅

    para cada (ação, novo_estado) em função_sucessora(problema, estado(nó)) faça
        filho ← criar_nó(novo_estado)

        pai(filho) ← nó
        ação(filho) ← ação

        custo(filho) ← custo(nó) + custo_passo(nó, ação, novo_estado)
        profundidade(filho) ← profundidade(nó) + 1
        sucessores ← sucessores ∪ {filho}

    retornar sucessores
```


````python
class Busca(Generic[Estado, Acao]):

    def executar_busca(self, problema: 'Problema[Estado, Acao]', tipo_busca: TipoBusca, verbose: bool = False) -> Optional[List[Tuple[Acao, Estado]]]:
        """
        Método principal para executar o algoritmo de busca.

        :param problema: A definição do problema de busca.
        :param tipo_busca: O tipo de busca a ser executado.
        :param verbose: Se True, imprime detalhes da fronteira a cada iteração.
        :return: solução do problema - lista de ações ou estados
        """

        # Cria o nó inicial do problema
        nodo_inicial = Nodo(
            pai=None,
            acao=None,
            estado=problema.estado_inicial,
            custo_caminho=0.0,
            profundidade=0
        )

        # Inicializa a fronteira com o tipo de busca especificado
        fronteira = Fronteira(tipo_busca)

        # Adiciona o nó inicial à fronteira
        fronteira.adicionar(nodo_inicial)

        # Conjunto para armazenar estados já visitados e evitar ciclos.
        visitados = {nodo_inicial.estado}

        self.iteracoes = 0 # Adiciona um contador de iterações

        # Continua a busca enquanto a fronteira não estiver vazia
        while not fronteira.esta_vazia():
            self.iteracoes += 1 # Incrementa o contador de iterações

            # Remove o próximo nó a ser expandido da fronteira
            nodo_atual = fronteira.remover()

            if verbose:
                print(f"\n--- Iteração {self.iteracoes} ---")
                print(f"Expandindo estado: {nodo_atual.estado}")
                print(f"Custo acumulado: {nodo_atual.custo_caminho}, Profundidade: {nodo_atual.profundidade}")
                print(f"Fronteira (antes da expansão e adição de novos): {[n.estado for n in fronteira._container]}")
                print(f"Tamanho da fronteira: {len(fronteira._container)}")
                print(f"Estados visitados: {len(visitados)}")

            # Verifica se o estado do nó atual é um objetivo
            if problema.eh_objetivo(nodo_atual.estado):
                if verbose:
                    print(f"\nOBJETIVO ENCONTRADO na iteração {self.iteracoes}!")
                return self._solucao(nodo_atual)

            # Expande o nó atual para gerar seus sucessores
            for sucessor_nodo in self._expandir(nodo_atual, problema):

                # Adiciona à fronteira apenas se o estado ainda não foi visitado
                if sucessor_nodo.estado not in visitados:
                    visitados.add(sucessor_nodo.estado)
                    fronteira.adicionar(sucessor_nodo)
                    if verbose:
                        print(f"  Adicionado à fronteira: {sucessor_nodo.estado}")

        if verbose:
            print("\nFRONTEIRA VAZIA. Nenhuma solução encontrada.")

        # Retorna None se nenhuma solução for encontrada
        return None


    def _expandir(self, nodo: Nodo, problema: 'Problema[Estado, Acao]') -> List[Nodo]:
        """
        Expande um dado nó, gerando seus nós sucessores.

        :param nodo: O nó a ser expandido.
        :param problema: A instância do Problema.
        :return: Uma lista de novos Nodos sucessores.
        """
        # Inicializa uma lista vazia para armazenar os nós sucessores
        sucessores = []

        # Itera sobre os sucessores do estado atual
        # A função sucessora vai retornar uma lista de tuplas com (Ação, Estado vizinho, Custo do nodo até o vizinho)
        for acao, novo_estado, custo_da_acao in problema.obter_sucessores(nodo.estado):

            # Calcula o custo acumulado do caminho até o novo estado
            novo_custo_caminho = problema.calcular_custo_caminho(nodo.custo_caminho, custo_da_acao)

            # Atualiza a profundidade do novo nodo
            nova_profundidade = nodo.profundidade + 1

            # Adiciona um novo Nodo à lista de sucessores
            sucessores.append(
                Nodo(
                    estado=novo_estado,               # O novo estado
                    pai=nodo,                         # O nó pai é o nó atual
                    acao=acao,                        # A ação que levou a este estado
                    custo_caminho=novo_custo_caminho, # O custo acumulado
                    profundidade=nova_profundidade    # A profundidade na árvore de busca
                )
            )

        # Retorna a lista de nós sucessores
        return sucessores


    def _solucao(self, nodo_objetivo: Nodo) -> List[Tuple[Acao, Estado]]: # Método privado para reconstruir o caminho
        """
        Reconstrói o caminho da solução a partir de um nó objetivo encontrado.

        :param nodo_objetivo: O nó que representa o estado objetivo alcançado.
        :return: Uma lista de tuplas (ação, estado) do início ao objetivo.
        """
        caminho = []

        # Começa pelo nodo objetivo
        atual = nodo_objetivo

        # Percorre a árvore de volta até o nó inicial (pai é None)
        while atual.pai is not None:
            caminho.append((atual.acao, atual.estado))
            atual = atual.pai

        # Adiciona a raiz
        caminho.append((None, atual.estado))

        # Inverte o caminho para que fique do início ao objetivo
        caminho.reverse()

        return caminho
````

# Problemas

## 1) Quebra cabeças de 8

Implementação do problema do Quebra-Cabeça de 8 peças usando a classe Problema como base. A subclasse ProblemaQuebraCabecasDeOito implementa a modelagem do problema para ser resolvido por estratégias de busca.

O estado é representado por uma tupla de 9 inteiros:

```
(1, 2, 3,
 4, 5, 6,
 7, 8, 0)
```
onde, 0 representa o espaço vazio.



### Modelagem do problema

````python
class ProblemaQuebraCabecasDeOito(Problema[Tuple[int, ...], str]):

    def __init__(self, estado_inicial: Tuple[int, ...]):
        super().__init__(estado_inicial)
        self.estado_objetivo = (1, 2, 3, 4, 5, 6, 7, 8, 0)

    # Verifica se chegou no estado final
    def eh_objetivo(self, estado: Tuple[int, ...]) -> bool:
        return estado == self.estado_objetivo


    # Implementa a lógica de gerar os nós sucessores
    def obter_sucessores(self, estado: Tuple[int, ...]) -> List[Tuple[str, Tuple[int, ...], float]]:
        sucessores = []
        idx_vazio = estado.index(0) # Encontra a posição do espaço vazio (0)

        # Mapeia o índice 1D para coordenadas 2D (linha, coluna)
        linha, coluna = divmod(idx_vazio, 3)

        # Possíveis movimentos: (delta_linha, delta_coluna, nome_acao)
        movimentos = [
            (-1, 0, "cima"),     # Mover para cima
            (1,  0, "baixo"),    # Mover para baixo
            (0, -1, "esquerda"), # Mover para a esquerda
            (0,  1, "direita"),  # Mover para a direita
        ]

        for dl, dc, nome_acao in movimentos:
            nova_linha, nova_coluna = linha + dl, coluna + dc

            # Verifica se o novo movimento está dentro dos limites do tabuleiro
            if 0 <= nova_linha < 3 and 0 <= nova_coluna < 3:
                novo_idx = nova_linha * 3 + nova_coluna

                # Cria uma lista a partir da tupla para facilitar a troca
                lista_estado = list(estado)

                # Troca o 0 com a peça adjacente
                peca_a_mover = lista_estado[novo_idx]
                lista_estado[idx_vazio], lista_estado[novo_idx] = lista_estado[novo_idx], lista_estado[idx_vazio]

                novo_estado = tuple(lista_estado)
                acao = f"Mover {peca_a_mover} para {nome_acao}"
                custo = 1.0

                sucessores.append((acao, novo_estado, custo))

        return sucessores


    # Custo do caminho eh a soma de todas as ações
    def calcular_custo_caminho(self, custo_acumulado_anterior: float, custo_da_acao_atual: float) -> float:
        return custo_acumulado_anterior + custo_da_acao_atual
````

### Execução do algoritmo de busca

````python
def imprimir_tabuleiro(estado: tuple):
    for i in range(3):
        print("|", end="")
        for j in range(3):
            peca = estado[i * 3 + j]
            print(f" {peca if peca != 0 else ' '} ", end="|")
        print()


# Exemplo de uso para o ProblemaOitoPuzzle
estado_inicial_exemplo = (0, 2, 3,
                          1, 5, 6,
                          4, 7, 8)

problema_8_puzzle = ProblemaQuebraCabecasDeOito(estado_inicial_exemplo)

# Cria uma instância da classe de busca
busca = Busca()

# Executa a busca em largura com verbose=False
solucao = busca.executar_busca(problema_8_puzzle, TipoBusca.BuscaEmLargura, verbose=False)

if solucao:
    print("\n--- Solução para o Problema do Quebra-Cabeça de 8 (Busca em Largura) ---")
    for i, (acao, estado) in enumerate(solucao):
        # O primeiro elemento da solução é o estado inicial (None, estado_inicial_tuple)
        if i == 0:
            print(f"Estado Inicial:")
            imprimir_tabuleiro(estado) # Access the 'estado' attribute of the Nodo object
        else:
            print(f"\nPasso {i}: {acao}")
            imprimir_tabuleiro(estado)

    print(f"\nNúmero de passos: {len(solucao) - 1}")
    print(f"Custo total: {len(solucao) - 1}") # Cada passo tem custo 1.0
else:
    print("\nNenhuma solução encontrada com Busca em Largura.")
````

\## 2) Problema do Saltador (Jumper Problem)

Implementação do problema do "Saltador" onde peças azuis (`A`) devem se mover para a direita e peças vermelhas (`V`) para a esquerda, utilizando espaços vazios (`_`). As peças podem se mover para um espaço vazio adjacente ou pular uma peça de cor oposta se houver um espaço vazio atrás dela na direção do movimento.

O estado é representado por uma tupla de caracteres, por exemplo:

```
('A', 'A', 'A', '_', '_', '_', 'V', 'V', 'V')
```


### Modelagem do problema

````python
class ProblemaSaltador(Problema[Tuple[str, ...], str]):

    def __init__(self, estado_inicial: Tuple[str, ...]):

        super().__init__(estado_inicial)

        # O estado objetivo é todas as 'V's à esquerda, vazios no meio e todas as 'A's à direita
        num_azuis = estado_inicial.count('A')
        num_vermelhos = estado_inicial.count('V')
        num_vazios = estado_inicial.count('_')
        self.estado_objetivo = tuple(['V'] * num_vermelhos + ['_'] * num_vazios + ['A'] * num_azuis)

    def eh_objetivo(self, estado: Tuple[str, ...]) -> bool:
        return estado == self.estado_objetivo

    def obter_sucessores(self, estado: Tuple[str, ...]) -> List[Tuple[str, Tuple[str, ...], float]]:
        sucessores = []
        tamanho_tabuleiro = len(estado)
        lista_estado = list(estado)

        indices_vazios = [i for i, x in enumerate(estado) if x == '_']

        for idx_vazio in indices_vazios:
            # Tentar mover 'A' (move para a direita)
            # Movimento adjacente
            if idx_vazio - 1 >= 0 and lista_estado[idx_vazio - 1] == 'A':
                novo_estado_lista = list(lista_estado)
                novo_estado_lista[idx_vazio], novo_estado_lista[idx_vazio - 1] = novo_estado_lista[idx_vazio - 1], novo_estado_lista[idx_vazio]
                sucessores.append((f"Mover A da posição {idx_vazio - 1} para {idx_vazio}", tuple(novo_estado_lista), 1.0))

            # Movimento de salto ('A' pula 'V')
            if idx_vazio - 2 >= 0 and lista_estado[idx_vazio - 2] == 'A' and lista_estado[idx_vazio - 1] == 'V':
                novo_estado_lista = list(lista_estado)
                novo_estado_lista[idx_vazio], novo_estado_lista[idx_vazio - 2] = novo_estado_lista[idx_vazio - 2], novo_estado_lista[idx_vazio]
                sucessores.append((f"Pular V com A da posição {idx_vazio - 2} para {idx_vazio}", tuple(novo_estado_lista), 1.0))

            # Tentar mover 'V' (move para a esquerda)
            # Movimento adjacente
            if idx_vazio + 1 < tamanho_tabuleiro and lista_estado[idx_vazio + 1] == 'V':
                novo_estado_lista = list(lista_estado)
                novo_estado_lista[idx_vazio], novo_estado_lista[idx_vazio + 1] = novo_estado_lista[idx_vazio + 1], novo_estado_lista[idx_vazio]
                sucessores.append((f"Mover V da posição {idx_vazio + 1} para {idx_vazio}", tuple(novo_estado_lista), 1.0))

            # Movimento de salto ('V' pula 'A')
            if idx_vazio + 2 < tamanho_tabuleiro and lista_estado[idx_vazio + 2] == 'V' and lista_estado[idx_vazio + 1] == 'A':
                novo_estado_lista = list(lista_estado)
                novo_estado_lista[idx_vazio], novo_estado_lista[idx_vazio + 2] = novo_estado_lista[idx_vazio + 2], novo_estado_lista[idx_vazio]
                sucessores.append((f"Pular A com V da posição {idx_vazio + 2} para {idx_vazio}", tuple(novo_estado_lista), 1.0))

        return sucessores

    def calcular_custo_caminho(self, custo_acumulado_anterior: float, custo_da_acao_atual: float) -> float:
        return custo_acumulado_anterior + custo_da_acao_atual
````

### Execução do algoritmo de busca

````python
def imprimir_tabuleiro_saltador(estado: tuple):
    mapeamento_emojis = {
        'A': '🔵',  # Azul
        'V': '🔴',  # Vermelho
        '_': '⬜'   # Vazio
    }
    print("|", end="")
    for peca in estado:
        print(f" {mapeamento_emojis.get(peca, peca)} ", end="|") # Usa o emoji mapeado ou a própria peça se não houver mapeamento
    print()
````

````python
# Exemplo de uso para o ProblemaSaltador
# A A A _ _ _ V V V
estado_inicial_saltador = ('A', 'A', 'A', '_', '_', '_', 'V', 'V', 'V')

problema_saltador = ProblemaSaltador(estado_inicial_saltador)

# Cria uma instância da classe de busca
busca_saltador = Busca()

print("Estado Inicial do Problema Saltador:")
imprimir_tabuleiro_saltador(estado_inicial_saltador)
print(f"Estado Objetivo: {problema_saltador.estado_objetivo}")

# Executa a busca em largura
solucao_saltador = busca_saltador.executar_busca(problema_saltador, TipoBusca.BuscaEmLargura, verbose=False)

if solucao_saltador:
    print("\n--- Solução para o Problema do Saltador (Busca em Largura) ---")
    for i, (acao, estado) in enumerate(solucao_saltador):
        if i == 0:
            print(f"Estado Inicial:")
            imprimir_tabuleiro_saltador(estado)
        else:
            print(f"\nPasso {i}: {acao}")
            imprimir_tabuleiro_saltador(estado)

    print(f"\nNúmero de passos: {len(solucao_saltador) - 1}")
    print(f"Custo total: {len(solucao_saltador) - 1}")  # Cada passo tem custo 1.0
    print(f"Total de iterações da busca: {busca_saltador.iteracoes}")
else:
    print("\nNenhuma solução encontrada com Busca em Largura para o Problema do Saltador.")
````
